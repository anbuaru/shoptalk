#Load into pandas (5 min)

```python
#glob.glob(...): Searches your directory for all files matching the pattern
# *.json.gz (which are compressed JSON files).
#sorted(...): Sorts the list of file paths alphabetically (e.g., from
#listings_0.json.gz to listings_f.json.gz) to ensure they are processed in a
#deterministic order.
#for f in ...: Initiates a loop, assigning each file path one by one to the variable

#gzip.open(f, ...): Opens the compressed .gz file directly without needing to
# extract it on your hard drive first.
#'rt': Opens the file in read mode as text (text mode instead of bytes).
#encoding='utf-8': Decodes the file contents using standard UTF-8 text encoding.
#with ... as fh: Uses Python's context manager to handle the file stream (fh),
#ensuring the file is safely closed when the block finishes, even if an error occurs.

#for line in fh: Since these metadata files are in "JSON Lines" format (where each
# line is a separate, valid JSON object), this iterates through the file line by line.
#json.loads(line): Converts (deserializes) each text line into a Python dictionary.
#rows.append(...): Adds the newly parsed dictionary to the rows list, which is
#later converted into a pandas DataFrame.

import gzip, json, glob
import pandas as pd

rows = []
for f in sorted(glob.glob(f'{DATA_DIR_SHOPTALK}/listings/metadata/*.json.gz')):
    with gzip.open(f, 'rt', encoding='utf-8') as fh:
        for line in fh:
            rows.append(json.loads(line))

df = pd.DataFrame(rows)
print(df.shape)          # expect roughly 147,700 products
print(df.columns.tolist())
df.head(3)
```

#Handle the multilingual fields
##Most text fields are lists of {language_tag, value} entries in many languages. Pull out the English values:

```python
def en_values(field):
  if not isinstance(field,list):
    return[]
  return [d['value'] for d in field
          if isinstance(d,dict) and str(d.get('language_tag','')).startswith('en') and d.get('value')]

def first_en(field):
  v=en_values(field)
  return v[0] if v else None

def product_type(field):
  if isinstance(field,list) and field and isinstance(field[0],dict):
    return field[0].get('value')
  return None

df['name_en']      = df['item_name'].apply(first_en)
df['brand_en']     = df['brand'].apply(first_en)
df['color_en']     = df['color'].apply(first_en)
df['desc_en']      = df['product_description'].apply(lambda x: ' '.join(en_values(x)))
df['bullets_en']   = df['bullet_point'].apply(lambda x: ' '.join(en_values(x)))
df['keywords_en']  = df['item_keywords'].apply(lambda x: ', '.join(en_values(x)))
df['ptype']        = df['product_type'].apply(product_type)
```
